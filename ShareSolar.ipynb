```python
import pandas as pd

data = {
    "household": ["A", "B", "C", "D"],
    "day1_kWh": [12, 8, 15, 10],
    "day2_kWh": [14, 9, 16, 11],
    "day3_kWh": [13, 7, 14, 12],
    "day4_kWh": [15, 10, 18, 9],
    "day5_kWh": [16, 11, 17, 10],
    "day6_kWh": [14, 9, 15, 11],
    "day7_kWh": [13, 8, 16, 12],
}
df = pd.DataFrame(data)
print(df)
```

```python
df["total_kWh"] = df.iloc[:,1:].sum(axis=1)
print(df[["household","total_kWh"]])
```

```python
rate_per_kWh = 5
df["savings"] = df["total_kWh"] * rate_per_kWh
print(df[["household","total_kWh","savings"]])
```

```python
import matplotlib.pyplot as plt

plt.bar(df["household"], df["total_kWh"], color="green")
plt.xlabel("Household")
plt.ylabel("Total kWh Generated")
plt.title("Community Solar Contribution")
plt.show()
```

```python
df.to_csv("solar_data.csv", index=False)
```
